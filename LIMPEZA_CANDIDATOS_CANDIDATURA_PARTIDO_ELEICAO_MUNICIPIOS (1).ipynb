### **TABELA DE CANDIDATOS**

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "sequencial": "sequencial_candidato",
    "titulo_eleitoral": "titulo_eleitoral",
    "id_eleicao": "id_eleicao",
    "numero_partido": "numero_partido",
    "id_municipio": "id_municipio",
    "cargo": "cargo",
    "numero": "numero_urna",
    "nome_urna": "nome_urna",
    "situacao": "situacao"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las
df_candidatos = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).copy()


# Visualizar o resultado
df_candidatos.head()
```

```python
# ==========================================
# ANÁLISE DE DUPLICATAS (CANDIDATURA)
# ==========================================

# 1. Identificar duplicatas na chave (sequencial_candidato)
duplicados_candidatura = df_candidatos.groupby('sequencial_candidato').size()
duplicados_acima_de_1  = duplicados_candidatura[duplicados_candidatura > 1].sort_values(ascending=False)

print(f"Total de sequencial_candidato com duplicidade: {len(duplicados_acima_de_1)}")
print("\nExemplo dos que mais se repetem (sequencial | Quantidade):")
print(duplicados_acima_de_1.head(20))
```

```python
# 2. Investigar um exemplo para verificar se as linhas se complementam
exemplo_seq = duplicados_acima_de_1.index[0]  # caso queira conferir outros, basta trocar o numero
df_candidatos[df_candidatos['sequencial_candidato'] == exemplo_seq]
```

```python
df_dup_completos = df_candidatos[df_candidatos['sequencial_candidato'].duplicated(keep=False)]

print("--- DUPLICADOS POR CARGO E SEQUENCIAL_CANDIDATO ---")
print(
    df_dup_completos
    .groupby([ 'sequencial_candidato','id_eleicao'])
    .size()
    .sort_values(ascending=False)
    .reset_index(name='quantidade')
    .head(20)
)
```

```python
df_dup = df_dup_completos \
    .groupby(['sequencial_candidato','id_eleicao','nome_urna', 'titulo_eleitoral']) \
    .size() \
    .sort_values(ascending=False) \
    .reset_index(name='quantidade') \
    .head(30)
df_dup
```

```python
# 2.2 Análise dos duplicados agrupados por nome_urna e sequencial_candidato
print("--- DUPLICADOS POR NOME_URNA E SEQUENCIAL_CANDIDATO ---")
print(
    df_dup_completos
    .groupby(['nome_urna', 'sequencial_candidato'])
    .size()
    .sort_values(ascending=False)
    .reset_index(name='quantidade')
    .head(20)
)
```

### **TABELA DE PESSOAS**

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "titulo_eleitoral": "titulo_eleitoral",
    "cpf": "cpf",
    "nome": "nome",
    "data_nascimento": "data_nascimento",
    "genero": "genero",
    "raca": "raca"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las
df_pessoa = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).copy()


# Visualizar o resultado
df_pessoa.head()
```

```python
# Faz o agrupamento, conta as repetições e filtra apenas os que aparecem mais de 1 vez
duplicados = df_pessoa.groupby('titulo_eleitoral').size()
duplicados_acima_de_1 = duplicados[duplicados > 1].sort_values(ascending=False)

# Exibe o resultado
print(f"Total de títulos eleitorais com duplicidade: {len(duplicados_acima_de_1)}")
print("\nExemplo dos títulos que mais se repetem (Título | Quantidade de vezes):")
print(duplicados_acima_de_1.head(20))
```

```python
# Pega o número do título que mais se repetiu no passo anterior e joga aqui para investigar
exemplo_titulo = duplicados_acima_de_1.index[0] # caso queira conferir outros, basta trocar o numero

# Mostra todas as linhas desse mesmo título no arquivo original
df_pessoa[df_pessoa['titulo_eleitoral'] == exemplo_titulo]
```

```python
# 1. Criar um DataFrame apenas com as pessoas que têm títulos repetidos
# (Pegamos todas as ocorrências delas, não apenas a primeira)
df_duplicados_completos = df_pessoa[df_pessoa['titulo_eleitoral'].duplicated(keep=False)]

# 2. Vamos ver quantas colunas estão vazias (nulas) antes e depois por título
print("--- ANÁLISE DE COMPLEMENTARIDADE ---")
print(f"Total de linhas com títulos repetidos para analisar: {len(df_duplicados_completos)}")

# 3. Contar quantos valores válidos existem por coluna para os títulos repetidos
valores_por_coluna = df_duplicados_completos.groupby('titulo_eleitoral').count()

# Se o valor mínimo de preenchimento for diferente do valor máximo para o mesmo título,
# significa que uma linha tem o dado e a outra não (elas se complementam!)
print("\nExemplo de como as colunas estão preenchidas nos títulos duplicados:")
print(valores_por_coluna.head(10))
```

```python
# Junta as linhas preenchendo os buracos com o primeiro valor válido que encontrar
df_pessoa_combinado = df_pessoa.groupby('titulo_eleitoral').first().reset_index()

# Validação após aglutinar
print(f"\nLinhas restantes após combinar os dados: {len(df_pessoa_combinado)}")
print(f"Duplicados restantes: {df_pessoa_combinado['titulo_eleitoral'].duplicated().sum()}")
```

```python
# ==========================================
# 1. DIAGNÓSTICO INICIAL
# ==========================================
nulos_antes      = df_pessoa['titulo_eleitoral'].isnull().sum()
duplicados_antes = df_pessoa['titulo_eleitoral'].duplicated().sum()

print("--- DIAGNÓSTICO INICIAL (PESSOA) ---")
print(f"Valores nulos encontrados  : {nulos_antes}")
print(f"Valores duplicados encontrados: {duplicados_antes}")
print(f"Total de linhas no arquivo original: {len(df_pessoa)}\n")

# ==========================================
# 2. LIMPEZA DOS DADOS
# ==========================================

# Passo A: Remover linhas onde a Chave Primária (titulo_eleitoral) é nula
df_pessoa_limpo = df_pessoa_combinado.dropna(subset=['titulo_eleitoral']).copy()

# Passo B: Remover o ".0" gerado quando o título é lido como float
df_pessoa_limpo['titulo_eleitoral'] = (
    df_pessoa_limpo['titulo_eleitoral']
    .astype(str)
    .str.split('.')
    .str[0]
    .str.strip()
)

# ==========================================
# 3. VALIDAÇÃO FINAL
# ==========================================
nulos_depois      = df_pessoa_limpo['titulo_eleitoral'].isnull().sum()
duplicados_depois = df_pessoa_limpo['titulo_eleitoral'].duplicated().sum()

print("--- APÓS A LIMPEZA ---")
print(f"Valores nulos restantes    : {nulos_depois} (Esperado: 0)")
print(f"Valores duplicados restantes: {duplicados_depois} (Esperado: 0)")
print(f"Total de pessoas únicas salvas: {len(df_pessoa_limpo)}\n")
```

```python
df_pessoa_limpo.to_csv('/content/Pessoas_Limpo.csv', index=False)
print("Sucesso! O arquivo 'Pessoas_Limpo.csv' foi gerado e está pronto para o banco.")
```

### **TABELA DE ELEICAO**

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "id_eleicao": "id_eleicao",
    "ano": "ano",
    "tipo_eleicao": "tipo_eleicao",
    "data_eleicao": "data_eleicao"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las

df_deferido_eleicao = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).drop_duplicates().copy()
df_deferido_eleicao.reset_index(inplace=True)
df_deferido_eleicao = df_deferido_eleicao.drop(columns=['index'])

# Visualizar o resultado
df_deferido_eleicao.head()
```

```python
# ==========================================
# ANÁLISE DE DUPLICATAS (ELEIÇÃO)
# ==========================================

# 1. Identificar duplicatas
duplicados_eleicao = df_deferido_eleicao.groupby('id_eleicao').size()
duplicados_acima_de_1 = duplicados_eleicao[duplicados_eleicao > 1].sort_values(ascending=False)

print(f"Total de id_eleicao com duplicidade: {len(duplicados_acima_de_1)}")
print("\nExemplo dos que mais se repetem (id_eleicao | Quantidade):")
print(duplicados_acima_de_1.head(20))
```

```python
# 2. Investigar um exemplo para verificar se as linhas se complementam
exemplo_id = duplicados_acima_de_1.index[0]  # caso queira conferir outros, basta trocar o numero
df_deferido_eleicao[df_deferido_eleicao['id_eleicao'] == exemplo_id]
```

```python
# 3. Análise de complementaridade entre as colunas das duplicatas
df_dup_completos = df_deferido_eleicao[df_deferido_eleicao['id_eleicao'].duplicated(keep=False)]

print("--- ANÁLISE DE COMPLEMENTARIDADE ---")
print(f"Total de linhas com id_eleicao repetido para analisar: {len(df_dup_completos)}")

valores_por_coluna = df_dup_completos.groupby('id_eleicao').count()
print("\nExemplo de como as colunas estão preenchidas nos duplicados:")
print(valores_por_coluna.head(10))
```

```python
# 4. Combinar aproveitando o primeiro valor válido de cada campo
df_eleicao_combinado = df_deferido_eleicao.groupby('id_eleicao').first().reset_index()

print(f"Linhas restantes após combinar os dados: {len(df_eleicao_combinado)}")
print(f"Duplicados restantes: {df_eleicao_combinado['id_eleicao'].duplicated().sum()}")
```

```python
# ==========================================
# 1. DIAGNÓSTICO INICIAL
# ==========================================
nulos_antes      = df_deferido_eleicao['id_eleicao'].isnull().sum()
duplicados_antes = df_deferido_eleicao['id_eleicao'].duplicated().sum()

print("--- DIAGNÓSTICO INICIAL (ELEIÇÃO) ---")
print(f"Valores nulos encontrados  : {nulos_antes}")
print(f"Valores duplicados encontrados: {duplicados_antes}")
print(f"Total de linhas no arquivo original: {len(df_deferido_eleicao)}\n")

# ==========================================
# 2. LIMPEZA DOS DADOS
# ==========================================

# Passo A: Remover linhas onde a Chave Primária (id_eleicao) é nula
df_eleicao_limpo = df_eleicao_combinado.dropna(subset=['id_eleicao']).copy()

# Passo B: Garantir que id_eleicao seja inteiro sem decimais
df_eleicao_limpo['id_eleicao'] = (
    df_eleicao_limpo['id_eleicao']
    .astype(str)
    .str.split('.')
    .str[0]
    .str.strip()
)

# ==========================================
# 3. VALIDAÇÃO FINAL
# ==========================================
nulos_depois      = df_eleicao_limpo['id_eleicao'].isnull().sum()
duplicados_depois = df_eleicao_limpo['id_eleicao'].duplicated().sum()

print("--- APÓS A LIMPEZA ---")
print(f"Valores nulos restantes    : {nulos_depois} (Esperado: 0)")
print(f"Valores duplicados restantes: {duplicados_depois} (Esperado: 0)")
print(f"Total de eleições únicas salvas: {len(df_eleicao_limpo)}\n")
```

```python
df_eleicao_limpo.to_csv('/content/Eleicao_Limpo.csv', index=False)
print("Sucesso! O arquivo 'Eleicao_Limpo.csv' foi gerado e está pronto para o banco.")
```
