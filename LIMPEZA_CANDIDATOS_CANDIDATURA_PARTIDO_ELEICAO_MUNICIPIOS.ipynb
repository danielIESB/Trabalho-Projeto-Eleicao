**Limpar base para conter apenas:**

*   deferido
*   deferido com recurso



```python
df_deferido = df[(df['situacao'] == 'deferido') | (df['situacao'] == 'deferido com recurso')]
```

```python
df_deferido.reset_index(inplace=True)
```

```python
df_deferido = df_deferido.drop(columns=['index'])
```

```python
df_deferido
```

```python
df_deferido[['numero', 'nome_urna','numero_partido','nome']]
```

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "sequencial": "sequencial_candidato",
    "titulo_eleitoral": "titulo_eleitoral",
    "id_eleicao": "id_eleicao",
    "numero_partido": "numero_partido",
    "id_municipio": "id_municipio",
    "cargo": "cargo",
    "numero": "numero_urna",
    "nome_urna": "nome_urna",
    "situacao": "situacao"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las
df_deferido_tabela = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).copy()


# Visualizar o resultado
df_deferido_tabela.head()
```

```python
df_deferido_tabela.to_csv('candidatos.csv', index=False)
```

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "titulo_eleitoral": "titulo_eleitoral",
    "cpf": "cpf",
    "nome": "nome",
    "data_nascimento": "data_nascimento",
    "genero": "genero",
    "raca": "raca"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las
df_deferido_pessoas = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).copy()


# Visualizar o resultado
df_deferido_pessoas.head()
```

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    "id_eleicao": "id_eleicao",
    "ano": "ano",
    "tipo_eleicao": "tipo_eleicao",
    "data_eleicao": "data_eleicao"
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las

df_deferido_eleicao = df_deferido[list(mapeamento.keys())].rename(columns=mapeamento).drop_duplicates().copy()
df_deferido_eleicao.reset_index(inplace=True)
df_deferido_eleicao = df_deferido_eleicao.drop(columns=['index'])

# Visualizar o resultado
df_deferido_eleicao.head()
```

```python
df_municipio = pd.read_csv('https://raw.githubusercontent.com/danielIESB/Trabalho-Projeto-Eleicao/331f19fb7362f45847f748289f2079447ccf82ee/municipio_tse_ibge.csv', sep=';', encoding='latin1')
```

```python
df_municipio.head()
```

```python
municipio_candidatos = df_deferido[['id_municipio']].drop_duplicates().copy()
municipio_candidatos.reset_index(inplace=True)
municipio_candidatos = municipio_candidatos.drop(columns=['index'])
municipio_candidatos.head()
```

```python
df_municipio['CD_MUNICIPIO_IBGE'] = df_municipio['CD_MUNICIPIO_IBGE'].astype(float)
joined_municipios = pd.merge(municipio_candidatos, df_municipio, left_on='id_municipio', right_on='CD_MUNICIPIO_IBGE', how='inner')
display(joined_municipios.head())
```

```python
joined_municipios = joined_municipios.drop(columns=[
  'DT_GERACAO',
  'HH_GERACAO',
  'CD_UF_TSE',
  'CD_UF_IBGE',
  'NM_UF',
  'CD_MUNICIPIO_IBGE',
  'NM_MUNICIPIO_IBGE'
])
joined_municipios.head()
```

```python
joined_municipios.to_csv('municipios.csv', index=False)
```
