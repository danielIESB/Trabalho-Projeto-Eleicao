```python
import pandas as pd
```

```python
df = pd.read_csv('/content/Despesas_Candidatos_Presidente_Governador.csv', engine='python', on_bad_lines='skip')
```

```python
df.head()
```

```python
#Acrescentei uma coluno com o tipo de "transação" para importação na tabela Receitas e Despesas no banco de dados.
df['Tipo'] = 'D'
```

```python
df.head()
```

```python
df.columns
```

```python
# 1. Ajuste para tabela do BD
mapeamento = {
    'sequencial_despesa': 'sequencial',
    'sequencial_candidato': 'sequencial_candidato',
    'cpf_cnpj_fornecedor': 'cpf_cnpj_doador',
    'data_despesa': 'dthora_transacao',
    'valor_despesa': 'valor_transacao',
    'origem_despesa': 'origem',
    'tipo_despesa': 'natureza',
    'especie_recurso': 'especie',
    'numero_documento': 'numero_documento',
    'descricao_despesa': 'descricao_despesa',
    'Tipo': 'tipo_transacao'
}

# 2. Filtrar apenas as colunas desejadas e renomeá-las
df_receita_despesa = df[list(mapeamento.keys())].rename(columns=mapeamento).copy()

# 3. Converter dthora_transacao para o formato datetime se necessário, ignorando erros
df_receita_despesa['dthora_transacao'] = pd.to_datetime(df_receita_despesa['dthora_transacao'], errors='coerce')

# Visualizar o resultado
df_receita_despesa.head()
```

```python
df_receita_despesa.to_csv('/content/receitas_despesas.csv', index=False)
```
